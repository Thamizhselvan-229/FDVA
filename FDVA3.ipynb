```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Create sample dataset
data = {
    'Name': ['Arun', 'Bala', 'Chitra', 'Divya', 'Arun'],
    'Age': [20, 21, np.nan, 22, 20],
    'Marks': [85, 78, 92, np.nan, 85],
    'City': ['Chennai', 'Madurai', 'Chennai', 'Coimbatore', 'Chennai']
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Fill missing values
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())

# Remove duplicates
df.drop_duplicates(inplace=True)

# Min-Max Normalization
scaler = MinMaxScaler()
df[['Age', 'Marks']] = scaler.fit_transform(df[['Age', 'Marks']])

print("\nCleaned Data:")
print(df)
```
