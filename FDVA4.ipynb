```python
import pandas as pd

data = {
    'Name': ['Arun', 'Bala', 'Chitra', 'Divya', 'Eshan'],
    'Age': [20, 21, 19, 22, 20],
    'Marks': [85, 78, 92, 88, 76],
    'City': ['Chennai', 'Madurai', 'Chennai', 'Coimbatore', 'Madurai']
}

df = pd.DataFrame(data)

# First 5 rows
print("First 5 rows:")
print(df.head())

# Last 5 rows
print("\nLast 5 rows:")
print(df.tail())

# Information
print("\nData Information:")
print(df.info())

# Descriptive statistics
print("\nDescriptive Statistics:")
print(df.describe())

# Columns
print("\nColumns:")
print(df.columns)

# Unique values
print("\nUnique Cities:")
print(df['City'].unique())

# Statistics
print("\nMarks Statistics:")
print("Mean:", df['Marks'].mean())
print("Median:", df['Marks'].median())
print("Mode:", df['Marks'].mode().values)
print("Range:", df['Marks'].max() - df['Marks'].min())
print("Variance:", df['Marks'].var())
print("Standard Deviation:", df['Marks'].std())
```
